## Question 3: Multi-line to Single-line FASTA Converter

This question requires converting a multi-line FASTA file to a single-line FASTA format.

### FASTA Format Background
- FASTA format is commonly used to represent nucleotide or protein sequences
- Each sequence begins with a header line starting with '>' character
- In multi-line FASTA, sequence data can be split across multiple lines
- In single-line FASTA, each sequence is contained on a single line after its header

```python
def convert_fasta_to_single_line(input_filename, output_filename):
    try:
        with open(input_filename, 'r') as infile, open(output_filename, 'w') as outfile:
            current_header = None
            current_sequence = []
            
            for line in infile:
                line = line.strip()
                
                if not line:
                    continue
                    
                if line.startswith('>'):
                    if current_header is not None:
                        outfile.write(current_header + '\n')
                        outfile.write(''.join(current_sequence) + '\n')
                        
                    current_header = line
                    current_sequence = []
                else:
                    current_sequence.append(line)
            
            if current_header is not None:
                outfile.write(current_header + '\n')
                outfile.write(''.join(current_sequence) + '\n')
                
        print(f"Successfully converted {input_filename} to single-line format.")
        print(f"Output saved to {output_filename}")
        return True
        
    except Exception as e:
        print(f"Error during conversion: {str(e)}")
        return False

input_fasta = "multiline_input.fasta"
output_fasta = "singleline_output.fasta"

convert_fasta_to_single_line(input_fasta, output_fasta)
```
