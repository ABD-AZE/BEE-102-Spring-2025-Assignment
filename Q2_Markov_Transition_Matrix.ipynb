## Question 2: Building Markov Transition Matrix

This question requires building a first-order Markov model transition matrix from a nucleotide sequence. 

### Markov Model Overview
- States: The four nucleotides (A, C, G, T)
- Transition probabilities: Probability of transitioning from one nucleotide to another
- The transition matrix shows P(next state | current state)

```python
import pandas as pd
import numpy as np

def build_markov_matrix(sequence):
    nucleotides = ['A', 'C', 'G', 'T']
    
    transition_counts = np.zeros((4, 4))
    
    nucleotide_to_idx = {nuc: idx for idx, nuc in enumerate(nucleotides)}
    
    for i in range(len(sequence) - 1):
        current = sequence[i]
        next_base = sequence[i + 1]

        if current not in nucleotide_to_idx or next_base not in nucleotide_to_idx:
            print(f"Warning: Skipping invalid nucleotide(s) '{current}' -> '{next_base}'")
            continue
            
        row = nucleotide_to_idx[current]
        col = nucleotide_to_idx[next_base]
        transition_counts[row, col] += 1
    
    transition_probs = np.zeros_like(transition_counts, dtype=float)
    for i in range(4):
        row_sum = np.sum(transition_counts[i, :])
        if row_sum > 0:
            transition_probs[i, :] = transition_counts[i, :] / row_sum
    
    markov_matrix = pd.DataFrame(transition_probs, 
                               index=nucleotides, 
                               columns=nucleotides)
    
    return markov_matrix

dna_sequence = "CTTAGTCAGTTCCATGCAGATGACTGCTGCATAACGTACTGATCGATCTGACTGATT"
transition_matrix = build_markov_matrix(dna_sequence)

print("Markov Transition Matrix:")
print("Rows represent current base, columns represent next base")
print("Each entry (i,j) is the probability P(next=j | current=i)")
print()
print(transition_matrix.round(4))
```
